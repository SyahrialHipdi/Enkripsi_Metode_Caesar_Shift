```python
KUNCI = 10
```

```python
def enkripsi(teks):
  hasil_enkripsi = ""
  for e in teks:
    hasil_enkripsi += chr((ord(e)+KUNCI-65)%26+65)
  return hasil_enkripsi

def dekripsi(teks):
  hasil_dekripsi = ""
  for e in teks:
    hasil_dekripsi += chr((ord(e)-KUNCI-65)%26+65)
  return hasil_dekripsi
```

```python
print(dekripsi("ABCDEFGHIJKLMNOPQRSTUVWXYZ"))
```
